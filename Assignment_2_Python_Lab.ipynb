```python
import time
import schedule
from plyer import notification

def send_break_alert(title, message):
    """Display a native desktop notification."""
    notification.notify(
        title=title,
        message=message,
        timeout=5
    )

def hydration_alert():
    send_break_alert(
        "Hydration Alert",
        "Take a moment to drink some water."
    )

def posture_check():
    send_break_alert(
        "Posture Check",
        "Sit upright, relax your shoulders, and adjust your screen."
    )


schedule.every(10).seconds.do(hydration_alert)
schedule.every(20).seconds.do(posture_check)

print("Reminder daemon is running. Press Ctrl+C to stop.")
print("Hydration: every 10 seconds (testing); Posture: every 20 seconds.")

try:
    while True:
        schedule.run_pending()
        time.sleep(1)
except KeyboardInterrupt:
    print("Reminder daemon stopped. Stay hydrated and take breaks!")
```

```python
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils import get_column_letter

workbook = Workbook()
sheet = workbook.active
sheet.title = "Semester_Results"

headers = [
    "Roll No",
    "Student Name",
    "Lab Marks (out of 50)",
    "Theory Marks (out of 50)",
    "Total Marks (out of 100)",
    "Result"
]
sheet.append(headers)

students = [
    (1, "Aarav Sharma", 42, 45),
    (2, "Diya Patel", 38, 41),
    (3, "Ishaan Mehta", 18, 17),
    (4, "Ananya Shah", 46, 44),
    (5, "Kabir Desai", 25, 30),
]

for roll_no, name, lab_marks, theory_marks in students:
    total = lab_marks + theory_marks
    result = "Pass" if total >= 40 else "Fail"
    sheet.append([roll_no, name, lab_marks, theory_marks, total, result])

header_fill = PatternFill(fill_type="solid", fgColor="D9D9D9")
for cell in sheet[1]:
    cell.font = Font(bold=True)
    cell.fill = header_fill
    cell.alignment = Alignment(horizontal="center", vertical="center", wrap_text=True)

summary_row = sheet.max_row + 1
sheet.cell(row=summary_row, column=1, value="Class Average")
sheet.cell(row=summary_row, column=5, value=f"=AVERAGE(E2:E{summary_row - 1})")
sheet.cell(row=summary_row, column=1).font = Font(bold=True)
sheet.cell(row=summary_row, column=5).font = Font(bold=True)
sheet.cell(row=summary_row, column=5).number_format = "0.00"

sheet.freeze_panes = "A2"
sheet.auto_filter.ref = f"A1:F{summary_row - 1}"
sheet.row_dimensions[1].height = 32
column_widths = [12, 22, 24, 26, 25, 14]
for column_index, width in enumerate(column_widths, start=1):
    sheet.column_dimensions[get_column_letter(column_index)].width = width

output_file = "AIML_Batch_Grades.xlsx"
workbook.save(output_file)
print(f"Workbook saved as {output_file}")
print(f"Class average formula: =AVERAGE(E2:E{summary_row - 1})")
```
